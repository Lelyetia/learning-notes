```python
# 利用np.ndarray()从列表对象创建数组
import numpy as np
a1D = np.array([1,2,3,4])
a2D = np.array([[1,2],[3,4]])
a3D = np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
print(a1D,a2D, a3D, sep='\n')
```

```python
# 你可以自定义数组的类型，但是要注意其是否符合要求
# np.array([127,128,129], dtype=np.int8)  
```

```python
# 同样，要注意溢出问题
a = np.array([2,3,4],dtype= np.uint32)
b = np.array([5,6,7], dtype=np.uint32)
c_unsigned = a-b
print(c_unsigned)  # 输出 [4294967293 4294967293 4294967293]，因为无符号整数溢出
c_signed32 = a - b.astype(np.int32)
print(c_signed32)  # 输出 [-3 -3 -3]，因为有符号整数不会溢出，而是会变为负数
```

```python
# 创建一维数组的函数：np.linespace and np.arange
a = np.arange(10)
b = np.arange(2,10,dtype=np.float64)
c = np.arange(2,3,0.1)
print(a,b,c, sep='\n')
print(np.linspace(1.,4.,6))  # 创建一个一维数组，包含6个元素，范围从1到4，间隔均匀
```

```python
# 二维数组创建函数：np.eye,np.diag,np.vanedr
# np.eye创建单位矩阵
a = np.eye(3)  # 创建一个3x3的单位矩阵
print(a)
b = np.eye(3,5)
print(b)  # 创建一个3x5的单位矩阵

# np.diag创建对角矩阵
c = np.diag([1,2,3])  # 创建一个对角矩阵，主对角线为[1,2,3]
print(c)
d = np.diag([1,2,3], k=1)  # 创建一个对角矩阵，主对角线上方第一条对角线为1,2,3
print(d)
a = np.array([[1,2],[3,4]]) # 提取已有矩阵的对角线元素
e = np.diag(a)
print(e)

# np.vander创建范德蒙德矩阵，有关于范德蒙德矩阵，可参阅：https://zh.wikipedia.org/wiki/%E8%8C%83%E5%BE%B7%E8%92%99%E7%9F%A9%E9%99%A3
a = np.vander(np.linspace(0,2,5),2)
print(a)
b = np.vander([1,2,3,4],2)
print(b)
c = np.vander((1,2,3,4),4)
print(c)
```

```python
# 通用ndarray创建函数:np.ones,np.zeros,randoms
# 使用np.zeros创建一个全为0的数组
a = np.zeros((2,3))
print(a)
b = np.zeros((2,3,2))
print(b)

# np.ones:创建全为1的数组
c = np.ones((2,3))
print(c)
d = np.ones((2,3,2))
print(d)

# random方法的结果default_rng会创建一个包含0-1的随机数数组
from numpy.random import default_rng
e = default_rng(42).random((2,3))
print(e)
f = default_rng(42).random((2,3,2))
print(f)

# np.indices创建一组数组，每个数组代表一个维度的变化
g = np.indices((3,3))
print(g)
```

```python
# 复制、合并或修改现有数组
# 当数组或其元素赋值给新变量时，必需显示的np.copy数组，否则该变量只是对原始数组的一个视图
import numpy as np
a = np.array([1,2,3,4,5,6])
b = a[:2]
b+=1
print('a = ',a,'b = ',b)

# 使用np.copy数组创建一个新的数组
a = np.array([1,2,3,4])
b = a[:2].copy()
b += 1
print('a = ',a,'b = ',b)

# 采用np.block将两个2*2数组合并成一个4*4数组
A = np.ones((2,2))
B = np.eye(2)
C = np.zeros((2,2))
D = np.diag((-3,4))
E = np.block([[A,B],[C,D]])
print(E)
```
